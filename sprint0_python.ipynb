```python
a = np.array([1, 2, 3])
print(a + 1)          # 전체에 1씩 더하기 (루프 없이!)
print(a * 2)
print(a + a)           # 같은 shape끼리는 원소별 연산

c = np.array([[1,2,3],[4,5,6]])   # 2x3
d = np.array([10, 20, 30])           # shape (3,)
print(c + d)             # shape 다른데도 계산됨 — 이게 브로드캐스팅
```

```python
e = np.array([[1,2],[3,4]])
f = np.array([[5,6],[7,8]])
print(e * f)     # 원소별 곱 (같은 자리끼리)
print(e @ f)      # 진짜 행렬곱 (선형대수의 그것)
```

```python
g = np.array([[1,2,3],[4,5,6]])
print(g.sum())            # 전체 합
print(g.sum(axis=0))     # 열끼리 합 (세로 방향)
print(g.sum(axis=1))     # 행끼리 합 (가로 방향)
print(g.mean(axis=0))
```

```python
class Dog:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def bark(self):
    print(f"{self.name}가 짖습니다! 나이: {self.age}살")

d = Dog("초코", 3)
d.bark()
print(d.name)
```

```python
import numpy as np

class GradeBook:
  def __init__(self, name):
    self.name = name
    self.scores = np.array([]) # 빈 배열로 시작

  def add_score(self, score):
    self.scores = np.append(self.scores, score)

  def average(self):
    return self.scores.mean()

  def summary(self):
    print(f"{self.name}: 점수={self.scores}, 평균={self.average():.1f}")
```

```python
gb = GradeBook("승민")
gb.add_score(85)
gb.add_score(92)
gb.add_score(78)
gb.summary()
```

```python
class Animal:
  def __init__(self, name):
    self.name = name
  def speak(self):
    print(f"{self.name}가 소리를 냅니다")

class Cat(Animal):  # Animal을 상속
  def speak(self):  # 오버라이드
    print(f"{self.name}가 야옹합니다")
```

```python
c = Cat("나비")
c.speak()
```
